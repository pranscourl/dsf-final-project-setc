## 2. Observations Across the Tables

- **Same structure:** all three tables have the same 22 columns (the target plus 21 health indicators), no missing values, and every column stored as `float64`.
- **Same respondents:** `diabetes_3categories` and `diabetes_2categories_unbalanced` have the same size (253,680 rows) and identical summary statistics for the shared columns. The diabetes class is 35,346 rows in both.
- **Prediabetes is labeled differently:** in `diabetes_2categories_unbalanced`, class 0 (218,334) equals no diabetes plus prediabetes in `diabetes_3categories` (213,703 + 4,631), so prediabetes is counted as "no diabetes" there.
- **Slightly more duplicates in the binary table:** 24,206 vs 23,899. This is consistent with prediabetes and no-diabetes rows becoming identical once their labels are collapsed.
- **The 50/50 table is balanced by sampling:** it has 35,346 rows per class, the same as the diabetes group in the other two tables.

## 3. Decision on the Target Labels

To keep the three tables consistent, we use the same rule for all of them:

| Class | Meaning |
|---|---|
| 0 | No diabetes (including diabetes only during pregnancy, and prediabetes) |
| 1 | Diabetes |

For `diabetes_3categories`, this means prediabetes (1) is grouped with no diabetes (0), and diabetes (2) becomes 1. The other two tables already follow this rule.
