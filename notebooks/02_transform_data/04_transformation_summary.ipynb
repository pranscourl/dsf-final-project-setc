# Transformation Summary

This notebook summarizes and compares the three transform notebooks. Each raw table was recoded where needed, cleaned of missing values and duplicates, converted to integer types, and loaded back into `diabetes_final_db` as a new `_clean` table. The raw tables were not changed.

## 1. Target Label Rule

All three tables now use the same 2-class rule:

| Class | Meaning |
|---|---|
| 0 | No diabetes (including diabetes only during pregnancy, and prediabetes) |
| 1 | Diabetes |

Only `diabetes_3categories` needed recoding: prediabetes (1) was grouped with no diabetes (0) and diabetes (2) became 1. This matches the other two tables, whose class counts show that they already treat prediabetes as "no diabetes".

## 3. Observations Across the Tables

- **Identical results for two tables:** after the recode, `diabetes_3categories_clean` and `diabetes_2categories_unbalanced_clean` have the same size and class counts, and the same number of duplicates was removed (24,206). This is expected, since they hold the same respondents once the labels follow the same rule.
- **Same diabetes group everywhere:** all three cleaned tables have 35,097 diabetes rows, so the diabetes group is the same across the datasets.
- **The 50/50 table is no longer exactly balanced:** duplicate removal shifted it to 49.18% / 50.82%, because most of the removed rows (1,386 of 1,635) were in class 0.
- **Duplicates were mostly in class 0:** in the full tables, about 11% of the no-diabetes rows were removed compared with 0.7% of the diabetes rows. As a result, diabetes prevalence rose from 13.93% to 15.29%.
- **Limitation:** the dataset has no respondent ID, so some removed "duplicates" may be different people who gave identical answers.
