## 5. Detailed Explanation of the Variables

The relationships mentioned below are what health research commonly reports. They are only background. Whether our data shows them is tested in the EDA and statistics notebooks. Because every answer is self-reported survey data, the relationships are associations, not proof of cause.

### 5.1 Target Variable

**`Diabetes_binary`**: whether the respondent has diabetes. This is what the model predicts.
- **Coding:** 0 = no diabetes, 1 = diabetes.
- **How it was built:** in the survey, people report whether a doctor told them they have diabetes. The original data had 3 categories (no diabetes or only during pregnancy, prediabetes, diabetes). In the transform phase, all three tables were given the same 2-class rule, so **prediabetes and diabetes only during pregnancy are inside class 0**.
- **Self-reported diagnosis:** someone who has diabetes but was never diagnosed is labeled 0. Some 0s may therefore be real diabetes cases.
- **Artificial 50/50 split:** the data was balanced by undersampling in the load phase. In the original survey data, about 14% of respondents had diabetes, so 50% is not the real prevalence.
- **Why it matters for the model:** in screening, missing a person who has diabetes is worse than flagging someone who turns out to be healthy. That is why the project focuses on **Recall**.

### 5.2 Health Conditions and Measurements

- **`HighBP`** (binary): told by a health professional that blood pressure is high. High blood pressure commonly occurs together with diabetes and is one of the stronger risk indicators.
- **`HighChol`** (binary): told by a health professional that blood cholesterol is high. Like blood pressure, it is commonly found together with diabetes. Because the answer depends on being told by a professional, it also reflects whether the person has been tested.
- **`BMI`** (continuous): weight divided by height squared. It is the only measured, non-coded variable in the group. Excess body weight is one of the best-known risk factors for type 2 diabetes. BMI is calculated from self-reported height and weight, so it can be slightly off. About 1.65% of rows have a BMI of 50 or more and the highest value is 98, which we kept (see `01_initial_data_check`).
- **`Stroke`** (binary): ever told by a health professional that the respondent had a stroke. It is a serious condition that is more common among people with diabetes. It is rare in this dataset, so it carries limited information on its own.
- **`HeartDiseaseorAttack`** (binary): ever had coronary heart disease or a heart attack. Heart disease and diabetes are closely linked. Like `Stroke`, it is a lifetime history, so the order in which the conditions appeared is unknown.

### 5.3 Lifestyle

- **`Smoker`** (binary): has smoked at least 100 cigarettes in their lifetime. This is a **lifetime** measure, not current smoking, so a person who quit years ago still counts as 1. Smoking is associated with a higher risk of type 2 diabetes.
- **`PhysActivity`** (binary): did any physical activity or exercise in the past 30 days, outside of their regular job. Regular activity is associated with a lower risk of diabetes. The question is a simple yes/no, so it says nothing about how much activity was done.
- **`Fruits`** (binary): eats fruit one or more times per day. A diet indicator. The link to diabetes in surveys is usually weak because the question measures only frequency.
- **`Veggies`** (binary): eats vegetables one or more times per day. It has the same limitation as `Fruits`.
- **`HvyAlcoholConsump`** (binary): heavy drinker (more than 14 drinks a week for men, more than 7 for women). The relationship with diabetes in surveys is not straightforward. Some people reduce or stop drinking after a health diagnosis, so heavy drinking may look *less* common among people with diabetes. Interpret results with care.

### 5.4 Healthcare Access and Screening

These variables describe a person's contact with the healthcare system, which affects whether a diabetes diagnosis would ever be made.

- **`CholCheck`** (binary): had a cholesterol check in the past 5 years. Most respondents answered yes, so the variable is very lopsided. It indicates regular contact with healthcare.
- **`AnyHealthcare`** (binary): has any kind of health coverage. Also very lopsided. Coverage makes testing and diagnosis more likely.
- **`NoDocbcCost`** (binary): could not see a doctor in the past 12 months because of cost. This marks a cost barrier to care, which can delay diagnosis.

Since the target is a *diagnosed* condition, these variables may be related to the label partly because they affect who gets diagnosed, not only who has the disease.

### 5.5 Self-Reported Health Status

- **`GenHlth`** (ordinal, 1 to 5): the respondent's own rating of their health. **Lower is better** (1 = excellent, 5 = poor). The numbers are ordered categories, not measurements, so the gap between 1 and 2 is not necessarily the same as the gap between 4 and 5. Poorer ratings are strongly associated with diabetes.
- **`MentHlth`** (count, 0 to 30): number of days of poor mental health in the past 30 days. Most respondents report 0 days, but a few report 30, so the distribution is heavily skewed. The link with diabetes is generally weaker than for physical health.
- **`PhysHlth`** (count, 0 to 30): number of days of poor physical health (illness and injury) in the past 30 days. It has the same skewed shape as `MentHlth`.
- **`DiffWalk`** (binary): serious difficulty walking or climbing stairs. It is associated with poorer health and with diabetes complications, and also increases with age.

### 5.6 Demographics and Socioeconomic Factors

- **`Sex`** (binary): 0 = female, 1 = male. Differences in diabetes prevalence between the sexes are usually small compared with the other variables.
- **`Age`** (ordinal, 1 to 13): age in 5-year groups, from 18 to 24 up to 80 or older. The numbers are group codes, not ages in years. Diabetes becomes more common with age, so this is one of the most informative variables. Within each group, the exact age is unknown.
- **`Education`** (ordinal, 1 to 6): highest education completed, from no schooling to college graduate. Higher education is associated with lower diabetes prevalence, partly through income, health knowledge and access to care.
- **`Income`** (ordinal, 1 to 8): household income level, from less than $10,000 to $75,000 or more. The top code groups everyone earning $75,000 or more, so differences among higher earners are not visible. Lower income is associated with higher diabetes prevalence.

`Education` and `Income` tend to move together, which we check later in the statistics notebook (VIF).
